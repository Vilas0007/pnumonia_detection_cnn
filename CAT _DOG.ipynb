```python
import tensorflow as tf
import keras
```

load image and image data preprocessing 


```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator


train_datagen=ImageDataGenerator(rescale=1/255
)

training_set=train_datagen.flow_from_directory(
    r"C:\Users\vilas\Downloads\catdogtraining_set",
    target_size=(128,128),
    batch_size=32,
    class_mode='binary'

)


```

```python
training_set.class_indices
```

test data


```python
test_datagen=ImageDataGenerator(rescale=1./255)
test_set=test_datagen.flow_from_directory(r"C:\Users\vilas\Downloads\catdogtest_set",
target_size=(128,128),
batch_size=32,
class_mode='binary',
shuffle=False
)




 

```

modeling convolutional network 


```python
from keras.models import Sequential
model=Sequential()
```

MAX POLLING


```python
# convolution
from keras.layers import Conv2D
model.add(Conv2D(input_shape=[128,128,3],
        filters=50,kernel_size=3,activation='relu'))
```

```python
from keras.layers import MaxPooling2D
model.add(MaxPooling2D(pool_size=2,strides=2))
```

flattening

```python
from keras.layers import Flatten
model.add(Flatten())
```

full connection 

```python
from keras.layers import Dense
# hidden layers with 128 neurons 
model.add(Dense(units=128,activation='relu'))
model.add(Dense(units=128,activation='relu'))

model.add(Dense(units=1,activation='sigmoid'))
```

Tunning cnn model with data and testing the model with test data

```python
model.compile(optimizer='adam',
loss='binary_crossentropy',
metrics=['accuracy'])
```

```python
model.fit(x=training_set,validation_data=test_set,epochs=20)
```

```python
#Evalution 


import numpy as np
from PIL import Image

test_image=Image.open(r"C:\Users\vilas\Downloads\catdogtraining_set\training_set\dogs\9001.jpg")
#data preprosessing
test_image=test_image.resize((128,128))
test_image=np.array(test_image)
test_image=np.expand_dims(test_image,axis=0)
#predictionn
result=model.predict(test_image)


# evalution
if result[0][0]> 0.5:
    print('dog')
else:
    print('cat')


```
