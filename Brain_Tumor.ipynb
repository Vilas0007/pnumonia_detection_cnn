```python
import tensorflow as tf
import keras



```

```python
# load image and preprocessing 

from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen=ImageDataGenerator(rescale=1/255)
 

training_set=train_datagen.flow_from_directory(r"C:\Users\vilas\Downloads\archive (1)\Dataset\Brain Tumor CT scan Images",
target_size=(64,64),
 
class_mode='binary')


training_set.class_indices
```

```python

test_datagen=ImageDataGenerator(rescale=1/255)
 

testing_set=train_datagen.flow_from_directory(r"C:\Users\vilas\Downloads\archive (1)\Dataset\Brain Tumor MRI images",
target_size=(64,64),
 
class_mode='binary')


testing_set.class_indices
```

```python
#model convolution data
from keras.models import Sequential
model=Sequential()
```

```python
from keras.layers import Conv2D
model.add(Conv2D(input_shape=[64,64,3],
filters=15,
kernel_size=3,
activation='relu'))
```

```python
from keras.layers import MaxPooling2D
model.add(MaxPooling2D(pool_size=2,strides=2))
```

```python
#flattening
from keras.layers import Flatten
model.add(Flatten())
```

```python
#full coneection 
from keras.layers import Dense
model.add(Dense(units=128,activation='relu'))
model.add(Dense(units=1,activation='sigmoid'))
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',
metrics=['accuracy'])
```

```python
model.fit(x=training_set,validation_data=testing_set,epochs=12)
```
