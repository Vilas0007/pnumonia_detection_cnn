```python
import tensorflow as tf
import keras
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen=ImageDataGenerator(rescale=1/255)
training_set=train_datagen.flow_from_directory(
    r"C:\Users\vilas\Downloads\chest_xray\train",
    target_size=(64,64),
    class_mode='binary'
)


```

```python
training_set.class_indices
```

```python
#test data
test_datagen=ImageDataGenerator(rescale=1/255)
testing_set=test_datagen.flow_from_directory(
    r"C:\Users\vilas\Downloads\chest_xray\test",
    target_size=(64,64),
    class_mode='binary'
)
```

```python
#model convolutional data
from keras.models import Sequential
model=Sequential()

```

```python
#max_pollunng
 #convolution 

from keras.layers import Conv2D
model.add(Conv2D(input_shape=[64,64,3],
      filters=12,
      kernel_size=3,
      activation='relu'))
      
```

```python
#pooling
from keras.layers import MaxPooling2D
model.add(MaxPooling2D(pool_size=2,strides=2))
```

```python
#flattining
from keras.layers import Flatten
model.add(Flatten())
```

```python
# full connection 
from keras.layers import Dense
model.add(Dense(units=128,activation='relu'))
model.add(Dense(units=1,activation='sigmoid'))


```

```python
model.compile(optimizer='adam',
loss='binary_crossentropy',
metrics=['accuracy'])
```

```python
model.fit(x=training_set,validation_data=testing_set,epochs=10)
```

```python
#eveluted  
import numpy as np
from PIL import Image
test_image=Image.open(r"C:\Users\vilas\Downloads\chest_xray\pred\NORMAL\NORMAL_4.jpeg").convert('RGB')
#data preprocessing 

test_image=test_image.resize((64,64))
test_image=np.array(test_image)/255.0
test_image=np.expand_dims(test_image,axis=0)

#prediction
result=model.predict(test_image)

print(result)
if result[0][0]>0.5:
    print('pnumonia')
else:
    print('normal')

```
